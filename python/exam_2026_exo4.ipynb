```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl


# définition de T, ici on choisit une chaîne
# à 3 états qui a une forte probabilité de rester
# dans l'état courant.

T = np.array([[.8,.1,.1],
              [.1,.8,.1],
              [.1,.1,.8]])
N = T.shape[0]

print("T=",T)
print()


# Simulation directe de la chaîne de Markov (sans contraintes):

def simul_markov(T,n,pi0):
    N = T.shape[0]
    x = np.zeros(n, dtype=int)
    x[0] = np.random.choice(N)
    for k in range(n-1):
        x[k+1] = np.random.choice(N,p=T[x[k],:])
    return x

n = 100

x = simul_markov(T, n, pi0=np.ones(N)/N)

print("Simulation de la chaîne de Markov:")
print(x)
print()

# Simulation avec contraintes (3 états égaux à 0) via Metropolis:

def mu(x,T,pi0):
   n = len(x)
   p = pi0[x[0]]
   for k in range(n-1):
      p *= T[x[k],x[k+1]]
   return p

def simul_Q(x,N):
   n = len(x)
   while True:
      y = x.copy()
      ij = np.random.choice(n,size=2,replace=False)
      y[ij] = np.random.choice(N,size=2)
      if np.sum(y!=x)==2 and np.sum(y==0)==3:
         break
   return y
    
def simul_markov_contrainte(T,n,pi0,npas=10000):
    N = T.shape[0]
    x = np.array([0,0,0]+[1]*(n-3))
    for k in range(npas):
        y = simul_Q(x,N)
        u = np.random.rand()
        if u < mu(y,T,pi0)/mu(x,T,pi0):                
            x = y
    return x

x = simul_markov_contrainte(T, n, pi0=np.ones(N)/N)

print("Simulation de la chaîne de Markov, sous la contrainte de passer 3 fois par 0:")
print(x)
```
